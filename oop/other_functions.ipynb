# Set functions

```python
m = np.array([1,2,3,4,5])
n = np.array([3,4,5,6,7])
```

```python
np.union1d(m,n)
```

```python
np.intersect1d(m,n)
```

```python
print(np.setdiff1d(m,n))
print(np.setdiff1d(n,m))
```

```python
np.setxor1d(m,n)
```

```python
# numpy.in1d() function test whether each element of a 1-D array is also present in a second array and 
# return a boolean array the same length as arr1 that is True where an element of arr1 is in arr2 and False otherwise.
```

```python
np.in1d(m,1)
```

```python
m[np.in1d(m,1)]
```
